```python
import tkinter as tk

fenster = tk.Tk()  # Erstellt das Hauptfenster.
fenster.title("Registrierung")  # Legt den Fenstertitel fest.
fenster.geometry("420x430")  # Legt die Größe des Fensters fest.

# Zeigt eine Überschrift im Fenster an.
tk.Label(fenster, text="Konto erstellen", font=("Arial", 20)).pack(pady=15)

# Erstes Eingabefeld für die E-Mail-Adresse.
tk.Label(fenster, text="E-Mail-Adresse:").pack()
e_mail_feld = tk.Entry(fenster, font=("Arial", 14), width=30)
e_mail_feld.pack(pady=5)

# Erstes Passwortfeld. show="*" zeigt Sternchen statt der Zeichen.
tk.Label(fenster, text="Passwort:").pack()
passwort_feld = tk.Entry(fenster, font=("Arial", 14), width=30, show="*")
passwort_feld.pack(pady=5)

# Zweites Passwortfeld, um die Eingabe zu vergleichen.
tk.Label(fenster, text="Passwort wiederholen:").pack()
passwort_wiederholen_feld = tk.Entry(
    fenster,
    font=("Arial", 14),
    width=30,
    show="*"
)
passwort_wiederholen_feld.pack(pady=5)

# Zeigt das Passwort in beiden Feldern lesbar an.
def passwort_anzeigen():
    passwort_feld.config(show="")
    passwort_wiederholen_feld.config(show="")

# Versteckt das Passwort in beiden Feldern wieder mit Sternchen.
def passwort_verbergen():
    passwort_feld.config(show="*")
    passwort_wiederholen_feld.config(show="*")

# Prüft, ob die Felder ausgefüllt sind und die Passwörter übereinstimmen.
def pruefen():
    e_mail = e_mail_feld.get().strip()
    passwort = passwort_feld.get()
    passwort_wiederholt = passwort_wiederholen_feld.get()

    if not e_mail or not passwort or not passwort_wiederholt:
        rueckmeldung.config(
            text="Bitte fülle alle Felder aus.",
            fg="red"
        )
    elif "@" not in e_mail or "." not in e_mail.split("@")[-1]:
        rueckmeldung.config(
            text="Bitte gib eine gültige E-Mail-Adresse ein.",
            fg="red"
        )
    elif passwort != passwort_wiederholt:
        rueckmeldung.config(
            text="Die Passwörter stimmen nicht überein.",
            fg="red"
        )
    else:
        rueckmeldung.config(
            text="E-Mail und Passwörter sind in Ordnung.",
            fg="green"
        )

# Knöpfe zum Anzeigen oder Verbergen der Passwörter.
tk.Button(
    fenster,
    text="Passwort anzeigen",
    command=passwort_anzeigen
).pack(pady=5)

tk.Button(
    fenster,
    text="Passwort verbergen",
    command=passwort_verbergen
).pack(pady=5)

# Startet die Prüfung, wenn der Knopf angeklickt wird.
tk.Button(
    fenster,
    text="Überprüfen",
    command=pruefen
).pack(pady=10)

# In diesem Label wird das Prüfergebnis angezeigt.
rueckmeldung = tk.Label(fenster, text="", font=("Arial", 12))
rueckmeldung.pack(pady=5)

fenster.mainloop()  # Startet das Fenster und wartet auf Eingaben.
```
